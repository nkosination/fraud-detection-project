```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns


```

```python
df = pd.read_csv("data/raw/creditcard.csv")
df.head()
```

### Observation
The dataset was loaded successfully into a pandas DataFrame. The first five rows show the transaction features, including Time, Amount, the anonymized variables V1–V28, and the target variable Class. Each row represents a single credit card transaction.


```python
df.shape
```

### Observation

The dataset contains 284,807 transactions and 31 columns. The columns consist of 30 input features and 1 target variable (Class). The large number of observations is beneficial for training a machine learning model.

```python
df.info()

```

### Observation
The dataset contains numerical variables only. All columns appear to have complete observations with no obvious missing values. The target variable Class is stored as an integer, while most features are stored as floating-point numbers.

```python
df.describe()
```

### Observation
The summary statistics provide information about the distribution of numerical features. Transaction amounts vary considerably, indicating a wide range of transaction values.

```python
df["Class"].value_counts()
```

### Observation
The majority of transactions are labelled as Class 0 (normal), while only a small number are labelled as Class 1 (fraudulent). This indicates that the dataset is highly imbalanced, which is a common challenge in fraud detection problems.

```python
fraud_ratio = df["Class"].value_counts() / len(df) * 100
fraud_ratio 
```

### Observation
Approximately 99.83% of transactions are normal, while only about 0.17% are fraudulent. The extreme imbalance means that accuracy alone will not be a reliable evaluation metric for model performance.

### Count Plot

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Class",data = df)
plt.title("Fraud vs Normal Transactions")
plt.show()
```

### Observation 
The visualization clearly shows the imbalance between normal and fraudulent transactions. Fraud cases represent only a tiny fraction of the dataset, confirming the need for specialized techniques such as resampling and alternative evaluation metrics.

### Amount Statistics by Class

```python
df.groupby("Class")["Amount"].describe()
```

### Observation 
Comparing transaction amounts between normal and fraudulent transactions helps determine whether fraud is associated with unusual transaction values. Any differences observed suggest that transaction amount may provide useful information for fraud detection, although it is unlikely to be sufficient on its own.


### Amount Distribution Plot

```python
plt.figure(figsize=(8,5))
sns.histplot(df[df['Class'] == 0]["Amount"] , bins =50 , label="Normal", alpha=0.6)
sns.histplot(df[df['Class'] == 1]["Amount"] , bins =50 , label="Fraud", alpha=0.6)
plt.legend()
plt.title("Transaction Amount: Fraud vs Normal")
plt.show()
```

### Observation 
The distributions of transaction amounts for normal and fraudulent transactions overlap considerably. This suggests that transaction amount alone cannot reliably distinguish fraud from legitimate transactions. Additional features will therefore be required for effective fraud detection.

### Conclusion


The exploratory data analysis revealed that the dataset is highly imbalanced, with fraudulent transactions representing only a very small proportion of the data. The dataset contains no obvious missing values and appears to have been preprocessed through feature transformation. While transaction amount shows some variation between classes, it is not sufficient on its own to identify fraudulent activity. Future steps will focus on data preprocessing, handling class imbalance, and building predictive machine learning models.
